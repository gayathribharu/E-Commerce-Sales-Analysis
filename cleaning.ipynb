```python
order_details = pd.read_csv("OrderDetails.csv")
```

```python
print("Shape:", order_details.shape)

print("\nMissing values:")
print(order_details.isnull().sum())

print("\nDuplicate rows:", order_details.duplicated().sum())

print("\nDuplicate OrderDetailID:",
      order_details["OrderDetailID"].duplicated().sum())
```

```python
print("\nInvalid Quantity:")
print((order_details["Quantity"] <= 0).sum())

print("\nInvalid UnitPrice:")
print((order_details["UnitPrice"] <= 0).sum())

print("\nInvalid DiscountPercent:")
print((order_details["DiscountPercent"] < 0).sum())
```

```python
median_qty = order_details["Quantity"].median()

order_details["Quantity"] = order_details["Quantity"].mask(
    order_details["Quantity"] <= 0,
    median_qty
)
```

```python
median_price = order_details["UnitPrice"].median()

order_details["UnitPrice"] = order_details["UnitPrice"].mask(
    order_details["UnitPrice"] <= 0,
    median_price
)
```

```python
order_details["DiscountPercent"] = order_details["DiscountPercent"].mask(
    order_details["DiscountPercent"] < 0,
    0
)
```

```python
print("Invalid Quantity:",
      (order_details["Quantity"] <= 0).sum())

print("Invalid UnitPrice:",
      (order_details["UnitPrice"] <= 0).sum())

print("Invalid Discount:",
      (order_details["DiscountPercent"] < 0).sum())
```

```python
print(order_details.isnull().sum())
```

```python
print("Duplicate OrderDetailID:",
      order_details["OrderDetailID"].duplicated().sum())
```

```python
order_details = order_details.drop_duplicates()
```

```python
print("Duplicate rows:",
      order_details.duplicated().sum())
```

```python
order_details = order_details.drop_duplicates(
    subset="OrderDetailID",
    keep="first"
)
```

```python
print("Duplicate OrderDetailID:",
      order_details["OrderDetailID"].duplicated().sum())
```

```python
invalid_orders = order_details[
    ~order_details["OrderID"].isin(orders["OrderID"])
]

print("Invalid OrderID:", len(invalid_orders))
```

```python
invalid_products = order_details[
    ~order_details["ProductID"].isin(products["ProductID"])
]

print("Invalid ProductID:", len(invalid_products))
```

```python
order_details = order_details[
    order_details["OrderID"].isin(orders["OrderID"])
]
```

```python
print("Invalid OrderID:",
      (~order_details["OrderID"].isin(orders["OrderID"])).sum())

print("Rows:", len(order_details))
```

```python
print(order_details.isnull().sum())
```

```python
order_details["DiscountPercent"] = order_details["DiscountPercent"].fillna(0)
```

```python
print("Missing DiscountPercent:",
      order_details["DiscountPercent"].isnull().sum())
```

```python
print(order_details.isnull().sum())
```

```python
order_details.to_csv("OrderDetails_Cleaned.csv", index=False)

print("OrderDetails_Cleaned.csv saved successfully!")
```

```python
returns = pd.read_csv("Returns.csv")
```

```python
print("Shape:", returns.shape)

print("\nMissing values:")
print(returns.isnull().sum())

print("\nDuplicate rows:", returns.duplicated().sum())

print("\nDuplicate ReturnID:",
      returns["ReturnID"].duplicated().sum())
```

```python
print("Invalid RefundAmount:",
      (returns["RefundAmount"] < 0).sum())
```

```python
print("Missing RefundAmount:",
      returns["RefundAmount"].isnull().sum())
```

```python
returns["RefundAmount"] = returns["RefundAmount"].fillna(0)
```

```python
returns.loc[returns["RefundAmount"] < 0, "RefundAmount"] = 0
```

```python
print("Invalid RefundAmount:",
      (returns["RefundAmount"] < 0).sum())

print("Missing RefundAmount:",
      returns["RefundAmount"].isnull().sum())
```

```python
returns = returns.drop_duplicates()

print("Duplicate rows:", returns.duplicated().sum())
```

```python
print("Duplicate ReturnID:",
      returns["ReturnID"].duplicated().sum())
```

```python
returns = returns.drop_duplicates(
    subset="ReturnID",
    keep="first"
)
```

```python
print("Rows:", len(returns))
print("Duplicate ReturnID:",
      returns["ReturnID"].duplicated().sum())

print("\nMissing values:")
print(returns.isnull().sum())
```

```python
invalid_return_orders = returns[
    ~returns["OrderID"].isin(orders["OrderID"])
]

print("Invalid OrderID:", len(invalid_return_orders))
```

```python
returns["ReturnDate"] = pd.to_datetime(
    returns["ReturnDate"],
    format="mixed",
    dayfirst=True,
    errors="coerce"
)

print("Missing ReturnDate:",
      returns["ReturnDate"].isnull().sum())
```

```python
returns = returns[
    returns["OrderID"].isin(orders["OrderID"])
]
```

```python
print("Invalid OrderID:",
      (~returns["OrderID"].isin(orders["OrderID"])).sum())

print("Rows:", len(returns))
```

```python
returns.to_csv("Returns_Cleaned.csv", index=False)

print("Returns_Cleaned.csv saved successfully!")
```
